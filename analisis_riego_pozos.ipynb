```python
#entrenamiento y evaluación para la region 1 y 2
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np
import pandas as pd

def train_and_evaluate(data):

    # Separar variables
    X = data.drop('product', axis=1)
    y = data['product']
    
    X_train, X_valid, y_train, y_valid = train_test_split(
        X, y, test_size=0.25, random_state=42
    )
    
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    predictions = model.predict(X_valid)
    
    mean_predicted = predictions.mean()
    rmse = np.sqrt(mean_squared_error(y_valid, predictions))
    
    results = pd.DataFrame({
        'actual': y_valid.reset_index(drop=True),
        'predicted': predictions
    })
    
    return mean_predicted, rmse, results
```

<div class="alert alert-block alert-success">
<b>Acierto o fortaleza</b> <a class="tocSkip"></a><br>
En la celda [13] encapsulaste el entrenamiento y la evaluación en una función. Esto mejora mantenibilidad y reduce duplicación, algo especialmente útil cuando comparas regiones con el mismo proceso. Como continuidad, podrías devolver también el modelo entrenado si más adelante quisieras inspeccionar coeficientes o reutilizarlo, pero tal como está, cumple muy bien el objetivo.
</div>

```python

mean_0, rmse_0, results_0 = train_and_evaluate(geo_0)
mean_1, rmse_1, results_1 = train_and_evaluate(geo_1)
mean_2, rmse_2, results_2 = train_and_evaluate(geo_2)

```

Conclusión: Preparación para el cálculo del beneficio

El análisis previo muestra que ninguna región supera el umbral mínimo de producción requerido (111.1 mil barriles por pozo) en promedio. Por lo tanto, la rentabilidad del proyecto no puede basarse en valores promedio regionales, sino en la correcta selección de los pozos con mayor potencial productivo.

Para calcular el beneficio correctamente, es necesario:
1.-Utilizar las predicciones generadas por el modelo de regresión lineal.
2.-Seleccionar los 200 pozos con mayor volumen estimado en cada región.
3.-Calcular el volumen total esperado sumando las reservas reales correspondientes a esos pozos seleccionados.
4.-Multiplicar ese volumen por el ingreso por unidad (4500 USD).
5.-Restar el presupuesto total invertido (100 millones USD).

Esto permite estimar el beneficio potencial bajo una estrategia de selección óptima, alineada con las condiciones del negocio.

Además, dado que el modelo tiene error (RMSE distinto de cero), el beneficio calculado representa una estimación y deberá complementarse con un análisis de riesgo mediante bootstrapping para evaluar la probabilidad de pérdidas.

La lógica financiera es:
𝐺anancia=(Volumen total x 45000)-100,000,000
Importante:
Seleccionamos por predicción, pero calculamos con el valor real (actual).

```python
def calculate_profit(results, top_n, revenue_per_unit, budget):
    
    # Ordenar por predicción descendente
    top_wells = results.sort_values(by='predicted', ascending=False).head(top_n)
    
    # Sumar volumen real de esos pozos
    total_volume = top_wells['actual'].sum()
    
    # Calcular ingreso
    revenue = total_volume * revenue_per_unit
    
    # Ganancia
    profit = revenue - budget
    
    return profit, total_volume
```

<div class="alert alert-block alert-success">
<b>Acierto o fortaleza</b> <a class="tocSkip"></a><br>
En la celda [19] tu función de ganancia respeta el principio correcto del proyecto: seleccionar por <code>predicted</code> pero calcular ingreso con <code>actual</code>. Eso evita “hacer trampa” usando el target para seleccionar y refleja el mundo real: decides con pronósticos, pero el dinero se gana/pierde con lo que realmente hay.
</div>

```python
profit_0, volume_0 = calculate_profit(results_0, 200, REVENUE_PER_UNIT, BUDGET)
profit_1, volume_1 = calculate_profit(results_1, 200, REVENUE_PER_UNIT, BUDGET)
profit_2, volume_2 = calculate_profit(results_2, 200, REVENUE_PER_UNIT, BUDGET)
```

Porque la región presenta, como ya dijimos:

Pero, sabemos que existe error (RMSE), y el negocio exige: Riesgo de pérdida inferior al 2.5%
Por lo tanto, aunque Región 0 lidera en ganancia estimada, la decisión final dependerá del análisis de riesgo mediante bootstrapping.

```python
def bootstrap_profit(results, n_samples, sample_size, top_n, revenue_per_unit, budget):
    
    profits = []
    
    for i in range(n_samples):
        
        # Muestra aleatoria con reemplazo
        sample = results.sample(n=sample_size, replace=True, random_state=None)
        
        # Seleccionar los mejores 200 por predicción
        top_wells = sample.sort_values(by='predicted', ascending=False).head(top_n)
        
        # Calcular volumen real
        total_volume = top_wells['actual'].sum()
        
        # Calcular ganancia
        revenue = total_volume * revenue_per_unit
        profit = revenue - budget
        
        profits.append(profit)
    
    return pd.Series(profits)
```

<div class="alert alert-block alert-success">
<b>Acierto o fortaleza</b> <a class="tocSkip"></a><br>
En la celda [27] implementaste bootstrapping de manera coherente con el planteamiento: muestreo con reemplazo, selección del top por predicción y cálculo de ganancia con valores reales. Esto es valioso porque en vez de quedarte con un único número de beneficio, obtienes una distribución, que es la base para hablar de riesgo e intervalos de confianza.
</div>

```python
profits_0 = bootstrap_profit(results_0, 1000, 500, 200, REVENUE_PER_UNIT, BUDGET)
profits_1 = bootstrap_profit(results_1, 1000, 500, 200, REVENUE_PER_UNIT, BUDGET)
profits_2 = bootstrap_profit(results_2, 1000, 500, 200, REVENUE_PER_UNIT, BUDGET)
```

```python
#Para cada región necesitamos:
#Beneficio promedio
#Intervalo de confianza 95%
#Riesgo de pérdida (probabilidad de ganancia negativa)

def analyze_bootstrap(profits):
    
    mean_profit = profits.mean()
    
    lower = profits.quantile(0.025)
    upper = profits.quantile(0.975)
    
    risk = (profits < 0).mean()
    
    return mean_profit, lower, upper, risk
```

```python
mean_0, lower_0, upper_0, risk_0 = analyze_bootstrap(profits_0)
mean_1, lower_1, upper_1, risk_1 = analyze_bootstrap(profits_1)
mean_2, lower_2, upper_2, risk_2 = analyze_bootstrap(profits_2)

print("Región 0:", mean_0, lower_0, upper_0, risk_0)
print("Región 1:", mean_1, lower_1, upper_1, risk_1)
print("Región 2:", mean_2, lower_2, upper_2, risk_2)
```
